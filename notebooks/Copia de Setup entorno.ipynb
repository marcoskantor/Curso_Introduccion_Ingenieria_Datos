# Setup del repositorio

```python
import os
import sys
import shutil
from google.colab import drive, userdata

# ============================================
# 1. Montar Google Drive
# ============================================
drive.mount('/content/drive')

# ============================================
# 2. DEFINIR RUTAS (AJUSTAR SI ES NECESARIO)
# ============================================
# Ruta base donde tienes la carpeta ITBA
DRIVE_PATH = '/content/drive/MyDrive/ITBA'

# Nombre del repositorio en GitHub (DEBE COINCIDIR)
REPO_NAME = 'Curso_Introduccion_Ingenieria_Datos'

# Nombre de tu carpeta local con los archivos organizados (dentro de ITBA)
# ⚠️ CAMBIA ESTE NOMBRE si tu carpeta se llama distinto
SOURCE_FOLDER = 'MiProyecto_IngDatos'

# Rutas completas
PROJECT_PATH = os.path.join(DRIVE_PATH, REPO_NAME)
SOURCE_BASE = os.path.join(DRIVE_PATH, SOURCE_FOLDER)

# ============================================
# 3. OBTENER CREDENCIALES (Secretos en Colab)
# ============================================
try:
    pat = userdata.get('GITHUB_PAT')
    username = userdata.get('USERNAME')
    if not pat or not username:
        raise ValueError("Faltan credenciales. Asegúrate de crear los secretos GITHUB_PAT y USERNAME en el panel 🔑 de Colab.")
except Exception as e:
    print(f"❌ Error con las credenciales: {e}")
    print("👉 Solución: Ve al panel izquierdo → 🔑 Secretos → agrega GITHUB_PAT (tu token) y USERNAME (tu usuario).")
    raise

# ============================================
# 4. CLONAR REPOSITORIO (si no existe)
# ============================================
%cd {DRIVE_PATH}

if not os.path.exists(PROJECT_PATH):
    print("🔄 Clonando repositorio desde GitHub...")
    !git clone https://{username}:{pat}@github.com/{username}/{REPO_NAME}.git
else:
    print("✅ El repositorio ya existe en Drive.")

# ============================================
# 5. MOVERSE AL REPO Y COPIAR ARCHIVOS
# ============================================
%cd {PROJECT_PATH}
print("📁 Contenido actual del repositorio:", os.listdir('.'))

# Copiar carpetas desde SOURCE_BASE al repositorio
if os.path.exists(SOURCE_BASE):
    for folder in ['data', 'src', 'notebooks']:
        src = os.path.join(SOURCE_BASE, folder)
        dst = os.path.join(PROJECT_PATH, folder)
        if os.path.exists(src):
            shutil.copytree(src, dst, dirs_exist_ok=True)
            print(f"✅ Copiado {folder}")
        else:
            print(f"⚠️ La carpeta {folder} no existe en origen, se omite.")
else:
    print(f"❌ NO se encontró {SOURCE_BASE}.")
    print("👉 Crea la carpeta con ese nombre dentro de ITBA y asegúrate de tener data/, src/, notebooks/")

# ============================================
# 6. VERIFICAR NOTEBOOK
# ============================================
notebooks_path = os.path.join(PROJECT_PATH, 'notebooks')
if os.path.exists(notebooks_path):
    archivos = os.listdir(notebooks_path)
    print(f"📓 Archivos en notebooks/: {archivos}")
    if any('Setup' in f for f in archivos):
        print("✅ ¡Se encontró un notebook con 'Setup' en el nombre!")
    else:
        print("⚠️ No se encontró 'Setup entorno.ipynb' en notebooks/. Verifica que esté subido.")

# ============================================
# 7. AGREGAR src AL PATH DE PYTHON
# ============================================
sys.path.append(os.path.join(PROJECT_PATH, 'src'))

# ============================================
# 8. CONFIGURAR GIT Y SUBIR CAMBIOS
# ⚠️ REEMPLAZA "tu_email@ejemplo.com" con tu correo real de GitHub
# ============================================
!git config --global user.name {username}
!git config --global user.email "tu_email@ejemplo.com"   # ← ¡CAMBIA ESTO!

# Agregar y commitear cambios
!git add .
!git commit -m "Estructura inicial: data, src, notebooks"

# Subir a GitHub (cambia 'main' por 'master' si tu rama principal se llama así)
!git push origin main   # ← Cambia a 'master' si es necesario

# ============================================
# 9. MOSTRAR ESTADO FINAL
# ============================================
!git status
print("\n🎉 ¡Todo listo! Revisa tu repositorio en GitHub.")
print("👉 Luego ve a Zenodo, activa el repo, crea un Release y obtén tu DOI.")
```
